## Simple Pytorch example to experiment with cell level profiling
- Enable 'Profiling with Nsight Systems' from NVIDIA Nsight extension in the Jupyterlab menu
- Restart Kernel when prompted, which will show up a Green run icon next to 'Interrupt Kernel' button in the toolbar
- Use 'Grey' play/run button to run a cell **without** profiling
- Use new 'Green' play/run button to run a call **with** profiling

***Please note**, the options in 'NVIDIA Nsight' extension menu will be active only when you are on a notebook.*

```python
import torch
import nvtx

# Create two matrices
A = torch.randn(3, 4).cuda()
B = torch.randn(4, 2).cuda()
```

```python
# Matrix multiplication
C = torch.mm(A, B)
```

```python
# Element-wise operations
D = A * 2
```

```python
nvtx.push_range("elmentwise operation - sum", domain="nvtx-custom-range")
E = A + 1
nvtx.pop_range(domain="nvtx-custom-range")
```

```python
# Print results
print(f"A shape: {A.shape}")
print(f"B shape: {B.shape}")
print(f"C shape: {C.shape}")
print(f"Matrix multiplication result:\n{C}")
```

## Everything in a cell. Profile to generate single report for all these operations


```python
import torch
import nvtx

# Create two matrices
A = torch.randn(3, 4).cuda()
B = torch.randn(4, 2).cuda()

# Matrix multiplication
C = torch.mm(A, B)
D = A * 2

nvtx.push_range("elmentwise operation - sum", domain="nvtx-custom-range")
E = A + 1
nvtx.pop_range(domain="nvtx-custom-range")

# Print results
print(f"A shape: {A.shape}")
print(f"B shape: {B.shape}")
print(f"C shape: {C.shape}")
print(f"Matrix multiplication result:\n{C}")
```
